```python
import glob
import time

# Vamos pegar todos os vídeos que estão em subpastas de videos  
video_files = glob.glob("./videos/**/*.mp4", recursive=True)
print(video_files)

# Processar cada vídeo
for video_file in video_files:
    print(f"\nProcessando vídeo: {video_file}")
    start_time = time.time()
    
    video_id = processor.process_video(video_file)
    
    end_time = time.time()
    processing_time = end_time - start_time
    print(f"Tempo de processamento: {processing_time:.2f} segundos ({processing_time/60:.2f} minutos)")
```
